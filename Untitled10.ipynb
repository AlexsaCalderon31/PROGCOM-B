```python
def mostrar_stock(flores):
    print("--- STOCK ACTUAL - Precio sin IVA ---")
    for t, c, p in zip(flores['tipo'], flores['cant'], flores['precios']):
        print(f"{t:<10} | Cant: {c} | ${p} + IVA")
    print("--------------------\n")
```

```python
def comprar_flores(flores, compra_user):
    mostrar_stock(flores)
    flor_opc = input("Seleccione la flor que desea comprar: ").lower().strip()

    if flor_opc not in flores['tipo']:
        print(f"La flor {flor_opc} no está disponible")
        return

    idx = flores['tipo'].index(flor_opc)
    flor_cant = int(input("Ingrese la cantidad que desea comprar: "))

    if not (0 < flor_cant <= flores['cant'][idx]):
        print("Cantidad no disponible")
        return

    compra_user['tipo'].append(flor_opc)
    compra_user['cantidad'].append(flor_cant)
    compra_user['precio'].append(flores['precios'][idx])
    flores['cant'][idx] -= flor_cant
    print(f"¡Se han añadido {flor_cant} {flor_opc}(s) al carrito!")

    if flores['cant'][idx] == 0:
        flores["tipo"].pop(idx)
        flores["cant"].pop(idx)
        flores["precios"].pop(idx)
```

```python
def calcular_totales(compra_user, IVA):
    subtotal = sum(c*p for c,p in zip(compra_user['cantidad'], compra_user['precio']))
    iva_valor = subtotal * IVA
    total = subtotal + iva_valor
    return subtotal, iva_valor, total
```

```python
def pagar(compra_user, IVA):
    if not compra_user["tipo"]:
        print("El carrito de compras está vacío.")
        return

    subtotal, iva_valor, total_pagar = calcular_totales(compra_user, IVA)

    print("\n** TABLA DE COMPRAS **")
    print(f"{'Flores':<10} | {'Cant':<5} | {'Unit':<8} | {'Subtotal'}")
    print("-" * 55)
    for tipo, cantidad, precio in zip(compra_user['tipo'], compra_user['cantidad'], compra_user['precio']):
        print(f"{tipo.capitalize():<10} | {cantidad:<5} | ${precio:<7} | ${cantidad*precio}")
    print("-" * 55)
    print(f"Subtotal sin IVA: ${subtotal}\nIVA (19%): ${iva_valor:.0f}\nTOTAL A PAGAR: ${total_pagar:.0f}\n")

    print("Seleccione el medio de pago: \n1. Efectivo\n2. Tarjeta de crédito\n3. Tarjeta de débito")
    pago_opc = input("Seleccione su opción: \n")

    if pago_opc in ["1", "3"]:
        plata = int(input(f"El total con IVA es ${total_pagar:.0f}. ¿Cuánto dinero tiene? "))
        if plata >= total_pagar:
            print(f"Pago exitoso. Cambio: ${plata - total_pagar:.0f}")
        else:
            print(f"¡No le alcanza! Tiene ${plata} y necesita ${total_pagar:.0f}")
    elif pago_opc == "2":
        cuotas = int(input("¿Número de cobros / cuotas? "))
        print(f"Se cobrará en {cuotas} cuotas de ${total_pagar/cuotas:.0f}")
```

```python
def generar_factura(compra_user, IVA):
    if not compra_user["tipo"]:
        print("El carrito de compras está vacío.")
        return

    subtotal, iva_valor, total = calcular_totales(compra_user, IVA)
    print(f"\n{'Flores':<12} | {'Cantidad':<8} | {'Unit':<8} | {'Subtotal'}")
    print("-" * 60)
    for tipo, cantidad, precio in zip(compra_user['tipo'], compra_user['cantidad'], compra_user['precio']):
        print(f"{tipo.capitalize():<12} | {cantidad:<8} | ${precio:<7} | ${precio*cantidad}")
    print("-" * 60)
    print(f"SUBTOTAL: ${subtotal}\nIVA 19%: ${iva_valor:.0f}\nTOTAL FACTURA: ${total:.0f}")
```

```python
def quitar_del_carrito(flores, compra_user):
    if not compra_user["tipo"]:
        print("Carrito vacío")
        return

    for t, c in zip(compra_user["tipo"], compra_user["cantidad"]):
        print(f"- {t}: {c}")

    quitar_flor = input("¿Qué flor desea quitar? ").lower().strip()
    if quitar_flor not in compra_user["tipo"]:
        print("Esa flor no está en el carrito")
        return

    idx_c = compra_user["tipo"].index(quitar_flor)
    cant_comprada = compra_user["cantidad"][idx_c]
    quitar_cant = int(input(f"Usted compró {cant_comprada} {quitar_flor}(s). ¿Cuántas desea quitar? "))

    if quitar_cant > cant_comprada:
        print(f"No puede quitar más de {cant_comprada}")
        return

    if quitar_flor in flores["tipo"]:
        flores["cant"][flores["tipo"].index(quitar_flor)] += quitar_cant
    else:
        flores["tipo"].append(quitar_flor)
        flores["cant"].append(quitar_cant)
        flores["precios"].append(compra_user["precio"][idx_c])

    compra_user["cantidad"][idx_c] -= quitar_cant
    if compra_user["cantidad"][idx_c] == 0:
        compra_user["tipo"].pop(idx_c)
        compra_user["cantidad"].pop(idx_c)
        compra_user["precio"].pop(idx_c)

    print(f"Se retiró {quitar_cant} {quitar_flor}(s)")
```

```python
def floristeria():
    flores = {
        "tipo": ["rosa", "tulipan", "girasol", "amapola", "lirio"],
        "cant": [10, 17, 3, 8, 5],
        "precios": [4000, 80000, 5000, 15000, 23000]
    }
    compra_user = {"tipo": [], "cantidad": [], "precio": []}
    IVA = 0.19

    print("Bienvenido a mi floristeria - Precios sin IVA")

    while True:
        if len(flores['tipo']) == 0:
            print("\n¡Ya no hay flores en inventario! Saliendo...")
            break

        print("\nQué desea hacer?\n1. Comprar flores\n2. Pagar\n3. Generar factura con IVA\n4. Quitar del carrito\n5. Salir")
        opc = input("Seleccione su opción: \n")

        match opc:
            case "1": comprar_flores(flores, compra_user)
            case "2": pagar(compra_user, IVA)
            case "3": generar_factura(compra_user, IVA)
            case "4": quitar_del_carrito(flores, compra_user)
            case "5":
                print("Gracias por emplear nuestros servicios. Feliz día")
                break
            case _: print('Número incorrecto')

# Ejecutar
floristeria()
```
