```python
vocal= "aeiou"
tilde= "áéíóú"
consonante= "bcdfghjklmnñpqrstvwxyz"
```

```python
#vocales se mueven una a la derecha
#tildes dos hacia la derecha
#consonante una hacia la izquierda
#hola....> guke
```

```python
mensaje= "hola"
```

```python
codificado= " "
for variable in mensaje:
  if variable in vocal:
    match variable:
      case "a":codificado=codificado+"e"
      case "e":codificado=codificado+"i"
      case "i":codificado=codificado+"o"
      case "o":codificado=codificado+"u"
      case "u":codificado=codificado+"a"
  elif variable in tilde:pass
  elif variable in consonante:
     codificado=codificado+consonante[consonante.index(variable)-1]
  else:
    codificado=codificado+variable

  for variable in mensaje:
    if variable in tilde:
      match variable:
        case "a":codificado=codificado+"e"
        case "e":codificado=codificado+"i"
        case "i":codificado=codificado+"o"
        case "o":codificado=codificado+"u"
        case "u":codificado=codificado+"a"
    elif variable in tilde:pass
    elif variable in consonante:
      codificado=codificado+consonante[consonante.index(variable)+2]
    else:
      codificado=codificado+variable
```
